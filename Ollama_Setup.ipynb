## Running Ollama
---

In order to use Ollama it needs to run as a service in background parallel to your scripts. Becasue Jupyter Notebooks is built to run code blocks in sequence this make it difficult to run two blocks at the same time. As a workaround we will create a service using subprocess in Python so it doesn't block any cell from running.

Service can be started by command `ollama serve`.

`time.sleep(5)` adds some delay to get the Ollama service up before downloading the model.

````python
import threading
import subprocess
import time

def run_ollama_serve():
  subprocess.Popen(["ollama", "serve"])

thread = threading.Thread(target=run_ollama_serve)
thread.start()
time.sleep(5)
````

````python
# Video Inferencer for Google Colab
# Based on original VideoInferencer class
# MODIFIED TO SEND FRAME BATCHES

import cv2
import time
import logging
import os
import base64
import io
import numpy as np
import IPython.display as display
from PIL import Image
from datetime import datetime
from google.colab import files
from google.colab.patches import cv2_imshow
from IPython.display import clear_output, HTML, display
import json # Added for parsing JSON responses

# Since Ollama doesn't run directly in Colab, we'll create a modified version
# that can use other APIs or local processing when running in Colab
try:
    import ollama
    OLLAMA_AVAILABLE = True
except ImportError:
    OLLAMA_AVAILABLE = False
    print("Ollama not available in Colab. Using alternative processing.")

# --- Assuming VideoMonitoringResponse structure for parsing ---
# You might need to adjust this based on the exact JSON output format
# of your chosen model when prompted for JSON.
from pydantic import BaseModel, Field
from typing import Literal

class VideoMonitoringResponse(BaseModel):
    primary_classification: Literal[
        "NORMAL", "ACCIDENT", "FIRE", "VIOLENCE", "THEFT", "DISTURBANCE",
        "MEDICAL EMERGENCY", "FIGHT", "OTHER", "SYSTEM ERROR"
    ] = Field(..., description="The primary classification of the event.")
    confidence_level: Literal["Low", "Medium", "High"] = Field(
        ..., description="The confidence level of the classification."
    )
    detailed_description: str = Field(
        ..., description="A clear explanation of what was detected."
    )
    recommended_action: str = Field(
        ..., description="A suggested response based on the severity."
    )
# --- End of assumed structure ---


class ColabBatchVideoInferencer: # Renamed for clarity
    def __init__(self, model_name="llava:13b", video_source=None, output_dir="logs", log_level=logging.INFO, frames_per_batch=10, sample_rate=5):
        """
        Initialize the Video Inferencer for Google Colab (Batch Processing)

        Args:
            model_name (str): Model name to use (e.g., 'llava:13b', 'gemma:2b - adjust as needed)
            video_source: Video source (uploaded file path)
            output_dir (str): Directory to store logs
            log_level: Logging level
            frames_per_batch (int): How many sampled frames to send in one request.
            sample_rate (int): Process every Nth frame from the video source.
        """
        self._setup_logging(output_dir, log_level)
        self.model_name = model_name
        self.running = False
        # Modified system prompt to emphasize sequence analysis
        self.system_prompt = """
You are an AI-powered city surveillance monitoring system. Your core function is to meticulously analyze a **batch of sequential images**, provided as an `image_bytes_list`, to **detect, accurately classify, and analyze unusual activities** occurring in urban environments. Each element in the list represents a frame captured at a consistent interval (e.g., 1 second). Your primary directive is to provide **objective, evidence-based assessments** based *only* on the visual information within this batch, with high reliability, avoiding speculation or hallucination.

**Core Task:** Identify and precisely differentiate various types of events based *only* on the visual information present in the provided `image_bytes_list`. Determine if the activity depicted across the sequence is **NORMAL** or fits into one of the defined **UNUSUAL EVENT** categories.

**Guiding Principles for Analysis:**

1.  **Evidence is Paramount:** Base *all* classifications and descriptions strictly on visible evidence within the provided `image_bytes_list`. **Do not infer actions, intentions, or outcomes not directly observed in the frames.**
2.  **Analyze the Full Sequence:** Evaluate changes, motion, and context across *all consecutive frames within the `image_bytes_list`*. A single frame might be misleading; the full sequence reveals the event's nature.
3.  **Objective Observation:** Focus on *what* is happening (actions, object interactions, environmental changes) rather than *why* (inferred motives).
4.  **Strict Categorization:** Use the defined categories precisely. If an event clearly fits a category based on visual evidence across the sequence, classify it. If evidence is ambiguous or insufficient across the batch, reflect this in the confidence level or use "OTHER" cautiously.
5.  **Default to Normal:** Assume activity across the batch is **NORMAL** unless clear, observable visual evidence within the sequence meets the criteria for one of the unusual event categories.
6.  **Differentiate Clearly:** Pay close attention to the defining features of each category to avoid misclassification (e.g., distinguish between a panicked crowd fleeing [PUBLIC DISTURBANCE] vs. individuals in a direct physical fight [VIOLENCE]).

**Detection Categories (Mutually Exclusive where applicable):**

1.  **NORMAL:** Standard, everyday activities depicted across the image sequence with no indicators of immediate safety or security threats. Includes regular traffic flow, pedestrians walking, people conversing peacefully, routine deliveries, etc.
2.  **ACCIDENT:** - (falling of people is not accident)
    *   **Criteria:** Requires clear visual evidence within the sequence of a collision involving vehicles, a vehicle striking structures/pedestrians, significant structural collapse, or an industrial mishap resulting in visible damage or potential injury.
    *   **Distinction:** Not just heavy traffic or a stalled vehicle, but an actual impact or collapse event visible within the frames.
3.  **FIRE:**
    *   **Criteria:** Requires direct visual evidence within the sequence of **visible flames** OR **dense, persistent smoke** clearly emanating from a specific source (vehicle, building, object) OR rapid, unexplained environmental changes strongly indicative of combustion (e.g., sudden charring developing across frames).
    *   **Distinction:** Not steam, dust, or fog. Must be indicative of uncontrolled burning visible in the images.
4.  **VIOLENCE:**
    *   **Criteria:** Requires clear visual evidence within the sequence of one or more individuals engaging in **physical aggression** against another person (e.g., hitting, kicking, pushing forcefully, grappling aggressively) OR the **clear brandishing or use of a weapon** (e.g., knife, bat, firearm pointed threateningly). Includes small-scale fights and larger assaults visible in the frames.
    *   **Distinction:** Not just arguing or agitated gestures. Requires physical contact or weapon presence/use visible in the sequence.
5.  **THEFT / VANDALISM:**
    *   **Criteria:** Requires clear visual evidence within the sequence of:
        *   **Theft:** Snatching property (bag, phone) from a person, breaking into a vehicle/building, shoplifting (if clearly visible act occurs within the frames).
        *   **Vandalism:** Deliberate act of damaging property (e.g., smashing windows, graffiti in progress, breaking structures) observed happening within the sequence.
        *   **Unauthorized Access:** Forcing entry into restricted areas observed within the sequence.
    *   **Distinction:** Must see the act of taking/damaging or clear indicators like broken windows *immediately* following suspicious behaviour near them within the provided frames. Don't assume theft just because someone is running unless the act of taking is also seen.
6.  **PUBLIC DISTURBANCE:**
    *   **Criteria:** Requires visual evidence within the sequence of:
        *   **Large-scale disorder:** Rioting, mob behavior, widespread panic (large groups running chaotically *away* from an unseen or seen threat across multiple frames).
        *   **Unlawful Gatherings:** Large, disruptive protests blocking major thoroughfares or becoming aggressive (widespread property damage, coordinated aggression *not* classified as individual VIOLENCE) visible across the sequence.
    *   **Distinction:** Focuses on collective, disruptive behavior of a significant group evolving or present across the frames. Differentiated from VIOLENCE (individual/small group physical acts) or NORMAL (peaceful large gatherings).
7.  **MEDICAL EMERGENCY:** (dont inculde this in vehicle accidents ,collisions etc in MEDICAL EMERGENCY)
    *   **Criteria:** Requires clear visual evidence within the sequence of an individual **collapsing suddenly and remaining unresponsive**, having a visible seizure, or suffering an obvious severe injury (e.g., heavy bleeding after a fall) where bystanders show clear concern or attempt aid within the frames.
    *   **Distinction:** Not someone simply sitting or lying down.
8.  **OTHER UNUSUAL EVENT:**
    *   **Criteria:** Use this category **sparingly**. Applies *only* when there is clear visual evidence across the sequence of a situation posing a potential safety/security risk that **does not fit any other category precisely**. Examples: A person dangerously climbing infrastructure, discovery of a suspicious unattended package in a high-traffic area, sudden flooding developing across frames, a dangerous animal loose.
    *   **Constraint:** Requires a concrete, observable anomaly visible in the `image_bytes_list`. **Do not use for vague "suspicious behavior."**

**Analysis Process:**

1.  **Ingest & Compare:** Analyze the entire `image_bytes_list`, comparing consecutive frames.
2.  **Identify Key Elements:** Note presence and behavior of people, vehicles, objects, and environmental conditions across the sequence.
3.  **Detect Anomalies & Changes:** Identify deviations from typical patterns or significant changes occurring between frames (sudden speed changes, erratic movement, congregation, smoke, impacts, aggressive postures).
4.  **Contextualize:** Use the full sequence in the `image_bytes_list` to understand the flow of events leading to, during, and potentially after the anomaly.
5.  **Evidence-Based Classification:** Match the observed, concrete evidence from the *entire sequence* against the criteria for each category. Classify as **NORMAL** if no criteria are met across the batch.
6.  **Assign Confidence:** Determine confidence based on the clarity, duration, and unambiguity of the visual evidence *within the provided batch*.

**Output Requirements:**

For the event depicted in the analyzed `image_bytes_list`:

1.  **Primary Classification:** (NORMAL, ACCIDENT, FIRE, VIOLENCE, THEFT / VANDALISM, PUBLIC DISTURBANCE, MEDICAL EMERGENCY, OTHER UNUSUAL EVENT) - Use *exact* category names.
2.  **Confidence Level:**
    *   **High:** Clear, unambiguous visual evidence directly matching category criteria persists or develops clearly across the relevant sequence within the batch.
    *   **Medium:** Visual evidence suggests a category, but is partially obscured, brief, or slightly ambiguous within the batch. Multiple indicators may point towards the classification.
    *   **Low:** Some visual indicators hint at a category, but the evidence within the batch is unclear, fleeting, or insufficient for a definite classification. Requires caution.
3.  **Detailed Description:** **Objectively describe *only what is visually observed* across the sequence in the `image_bytes_list`.** Detail the actions, objects involved, location specifics (if discernible), and environmental changes observed across the frames. **Crucially, state *why* the visual evidence supports the chosen classification**, referencing specific category criteria observed developing or occurring within the sequence. (e.g., "The image sequence shows a red car impacting a stationary blue van between frames 3 and 4 [ACCIDENT criterion]. Debris is visible scattering in subsequent frames. No individuals appear injured." OR "Frames 2 through 7 show two individuals exchanging punches and grappling on the ground [VIOLENCE criterion]. Bystanders are observed moving away.")
4.  **Recommended Action:** Suggest a response based on classification and confidence. (e.g., "Alert emergency services (Police/Fire/Medical)," "Dispatch security personnel for assessment," "Monitor situation closely for escalation," "Log event, no immediate action needed").

**Example Outputs (Maintain format):**

1.  **Primary Classification:** FIRE
    **Confidence Level:** High
    **Description:** The sequence in `image_bytes_list` shows dense black smoke persistently rising from the engine compartment of a stationary white sedan, starting faintly in early frames and intensifying over the sequence. Small flames become visible near the front wheel in later frames (e.g., frame 8 onwards). Several people are seen moving quickly away from the vehicle across multiple frames. Evidence matches FIRE criteria (persistent smoke development, visible flames).
    **Recommended Action:** Notify the fire department immediately and alert police.

2.  **Primary Classification:** VIOLENCE
    **Confidence Level:** Medium
    **Description:** Across the image sequence, two individuals are observed in a heated argument (frames 1-3). In frame 4, one individual forcefully shoves the other, who stumbles back. The altercation continues with aggressive posturing in subsequent frames, but clear punches are obscured by distance/angle. Meets VIOLENCE criteria (physical aggression - shoving observed). Confidence is Medium due to partial obscuration of subsequent actions within the batch.
    **Recommended Action:** Dispatch security personnel to investigate and de-escalate. Monitor closely.

3.  **Primary Classification:** NORMAL
    **Confidence Level:** High
    **Description:** The sequence provided in `image_bytes_list` consistently shows pedestrians walking on the sidewalk, vehicles moving at normal speed in traffic lanes, and people waiting at a bus stop across all frames. No observed actions meet criteria for any unusual event categories.
    **Recommended Action:** No action required. Continue monitoring (or process next batch).

        """
        self.last_prediction = "INITIALIZING"
        self.last_confidence = "Low"
        self.last_details = "Awaiting analysis"
        self.last_action = "None"

        self.video_source = video_source
        self.cap = None
        self.frame_count = 0
        self.total_frames_in_video = 0
        self.fps = 30
        self.sample_rate = max(1, sample_rate) # Process every Nth frame
        self.frames_per_batch = frames_per_batch # How many frames per API call

        self.last_process_time = 0
        self.processing_fps = 0 # Less meaningful in batch mode, represents read speed

        self.display_height = 480
        self.frame_buffer = [] # Buffer to hold frames for batch processing
        self.output_frames = [] # For creating animation/video in Colab
        self.response_history = {} # Store responses per batch

        if OLLAMA_AVAILABLE:
            # Increase timeout for potentially longer batch processing
            self.client = ollama.Client(timeout=120)
            self.logger.info(f"Using Ollama with model: {model_name}")
            # self.warm_up() # Warmup might take longer with sequence prompts
        else:
            self.client = None
            self.logger.info(f"Using alternative processing method for Colab (Simulation)")

        self.logger.info(f"ColabBatchVideoInferencer initialized (Batch Size: {self.frames_per_batch}, Sample Rate: {self.sample_rate})")

    def _setup_logging(self, output_dir, log_level):
        """Set up logging configuration"""
        if not os.path.exists(output_dir):
            os.makedirs(output_dir)
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
        log_file = os.path.join(output_dir, f"colab_batch_inferencer_{timestamp}.log")
        self.logger = logging.getLogger('ColabBatchVideoInferencer')
        self.logger.setLevel(log_level)
        if self.logger.hasHandlers(): # Prevent duplicate handlers if re-initialized
            self.logger.handlers.clear()
        file_handler = logging.FileHandler(log_file)
        file_formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')
        file_handler.setFormatter(file_formatter)
        console_handler = logging.StreamHandler()
        console_formatter = logging.Formatter('%(asctime)s - %(levelname)s - %(message)s')
        console_handler.setFormatter(console_formatter)
        self.logger.addHandler(file_handler)
        self.logger.addHandler(console_handler)

    def upload_video(self):
        """Upload a video file in Colab"""
        try:
            print("Please upload a short video file:")
            uploaded = files.upload()
            if not uploaded:
                self.logger.warning("No file was uploaded")
                return False
            filename = next(iter(uploaded))
            self.video_source = filename
            self.logger.info(f"Video uploaded: {filename}")
            # Reset state if a new video is uploaded
            self.frame_count = 0
            self.output_frames = []
            self.frame_buffer = []
            self.response_history = {}
            self.last_prediction = "INITIALIZING"
            return True
        except Exception as e:
            self.logger.error(f"Error uploading video: {str(e)}")
            return False

    # def warm_up(self): # Optional: Warmup might need adjustment for sequence prompts
    #     if not OLLAMA_AVAILABLE: return
    #     start_time = time.time()
    #     try:
    #         # Use a simple prompt, potentially with a dummy image if needed by model
    #         response = self.client.generate(
    #                       model=self.model_name,
    #                       prompt="System ready.",
    #                       keep_alive=-1, # Keep model loaded
    #                   )
    #         response_text = response.get('response', 'No response text')
    #         model_warmup_time = time.time() - start_time
    #         self.logger.info(f"Warmed up {self.model_name} in {model_warmup_time:.2f}s: {response_text}")
    #     except Exception as e:
    #         self.logger.error(f"Error during model warm-up: {e}")


    def start(self):
        """Start the video inferencer"""
        if self.running:
            self.logger.warning("Video inferencer is already running")
            return False
        if not self.video_source:
            self.logger.error("No video source specified. Please upload a video first.")
            return False

        try:
            self.cap = cv2.VideoCapture(self.video_source)
            if not self.cap.isOpened():
                self.logger.error(f"Failed to open video source: {self.video_source}")
                return False

            self.fps = self.cap.get(cv2.CAP_PROP_FPS) or 30
            self.total_frames_in_video = int(self.cap.get(cv2.CAP_PROP_FRAME_COUNT))
            width = int(self.cap.get(cv2.CAP_PROP_FRAME_WIDTH))
            height = int(self.cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
            self.logger.info(f"Video FPS: {self.fps:.2f}, Total Frames: {self.total_frames_in_video}, Resolution: {width}x{height}")

            self.running = True
            self.frame_count = 0
            self.frame_buffer = []
            self.output_frames = [] # Clear previous outputs
            self.logger.info("Video inferencer started - Reading frames...")
            self._process_video() # This will now read all frames then process
            return True

        except Exception as e:
            self.logger.exception(f"Error starting video inferencer: {str(e)}") # Use logger.exception for stack trace
            self.stop()
            return False

    def stop(self):
        """Stop the video inferencer and release resources"""
        self.running = False
        if self.cap is not None:
            self.cap.release()
            self.cap = None # Ensure it's None after release
        self.logger.info("Video inferencer stopped")
        # Display results AFTER processing is complete
        if len(self.output_frames) > 0:
             # Apply the final overlay before displaying
            self._apply_final_overlay_to_output()
            self._display_results()
        else:
             self.logger.info("No frames were processed or stored for display.")


    def _process_video(self):
        """Read video frames, sample them, and trigger batch processing at the end."""
        self.logger.info("Starting video frame reading loop...")
        progress_html = HTML('<h3>Reading video frames...</h3>')
        display(progress_html)
        frames_read = 0
        sampled_frames_count = 0

        try:
            while self.running and self.cap and self.cap.isOpened():
                read_start_time = time.time()
                ret, frame = self.cap.read()
                if not ret:
                    self.logger.info("End of video stream reached.")
                    break

                self.frame_count += 1
                frames_read += 1

                # Sample frames based on sample_rate
                if self.frame_count % self.sample_rate == 0:
                    sampled_frames_count += 1
                    # Store the original frame (or a resized copy for memory)
                    # If memory is a concern, resize here:
                    # resized_frame = self._resize_frame(frame, for_processing=True)
                    # self.frame_buffer.append(resized_frame)
                    self.frame_buffer.append(frame.copy()) # Store copy

                    # Also store a display version (can be resized differently)
                    display_frame = self._resize_frame(frame, for_display=True)
                    # Add basic info overlay during reading (optional)
                    self._add_reading_info_overlay(display_frame, self.frame_count)
                    self.output_frames.append(display_frame) # For final animation

                    # Optional: Display progress intermittently
                    if sampled_frames_count % 10 == 0:
                       clear_output(wait=True)
                       cv2_imshow(display_frame)
                       progress_html.value = f'<h3>Reading video... Frame {self.frame_count}/{self.total_frames_in_video} (Sampled: {sampled_frames_count})</h3>'
                       display(progress_html)


                read_time = time.time() - read_start_time
                self.processing_fps = 1.0 / max(read_time, 0.001) # Reading FPS

            self.logger.info(f"Finished reading video. Total frames read: {frames_read}. Total frames sampled: {len(self.frame_buffer)}")

            # --- Process the collected frames in batches ---
            if self.frame_buffer:
                progress_html.value = f'<h3>Processing {len(self.frame_buffer)} sampled frames...</h3>'
                display(progress_html)
                self._process_frame_batch(self.frame_buffer)
            else:
                self.logger.warning("No frames were sampled or buffered for processing.")
                self.last_prediction = "NORMAL"
                self.last_confidence = "Low"
                self.last_details = "No frames processed."
                self.last_action = "None"


        except Exception as e:
            self.logger.exception(f"Error during video processing loop: {str(e)}")
        finally:
            # Ensure stop is called to release resources and display results
            self.stop()


    def _resize_frame(self, frame, for_display=False, for_processing=False):
        """Resize frame, optionally differentiating for display vs processing"""
        # Simple resize for display based on height limit
        if for_display:
            height, width = frame.shape[:2]
            if height > self.display_height:
                scale = self.display_height / height
                new_width = int(width * scale)
                return cv2.resize(frame, (new_width, self.display_height), interpolation=cv2.INTER_AREA)
            return frame
        # Add resizing logic for processing if needed (e.g., model input limits)
        elif for_processing:
             # Example: resize to a fixed size for the model if necessary
             # return cv2.resize(frame, (512, 512), interpolation=cv2.INTER_AREA)
             return frame # Default: use original for processing buffer
        return frame # Default


    def _process_frame_batch(self, frame_batch):
        """Process a batch of frames with the LLM."""
        if not frame_batch:
            self.logger.warning("Attempted to process an empty frame batch.")
            return

        self.logger.info(f"Processing batch of {len(frame_batch)} frames.")
        image_bytes_list = []
        try:
            # Encode all frames in the batch
            for i, frame in enumerate(frame_batch):
                success, encoded_image = cv2.imencode('.jpg', frame)
                if success:
                    image_bytes_list.append(encoded_image.tobytes())
                else:
                    self.logger.error(f"Failed to encode frame index {i} in batch.")
                    # Decide: skip frame or abort batch? Let's skip for now.
                    continue

            if not image_bytes_list:
                self.logger.error("No frames could be encoded in the batch.")
                self.last_prediction = "SYSTEM ERROR"
                self.last_details = "Frame encoding failed for batch."
                return

            prompt = "Analyze the following sequence of frames and provide your assessment based on the instructions."
            start_time = time.time()
            response_text = None

            if OLLAMA_AVAILABLE and self.client:
                self.logger.info(f"Sending {len(image_bytes_list)} images to Ollama model {self.model_name}...")
                try:
                    response = self.client.generate(
                        model=self.model_name,
                        prompt=prompt,
                        images=image_bytes_list, # Send list of bytes
                        system=self.system_prompt,
                        format = VideoMonitoringResponse.model_json_schema(),
                        keep_alive=-1,
                        options={"temperature": 0.91} # Adjust temperature if needed
                    )
                    response_text = response.response

                except Exception as e:
                    self.logger.error(f"Ollama API call failed: {e}")
                    response_text = f'{{"primary_classification": "SYSTEM ERROR", "confidence_level": "Low", "detailed_description": "Ollama API error: {e}", "recommended_action": "Check Ollama service and connection."}}'


            else:
                # Simulate response for the entire batch
                self.logger.info("Simulating AI response for batch (Ollama not available)...")
                response_text = self._simulate_batch_ai_response(frame_batch)

            inference_time = time.time() - start_time
            self.last_process_time = inference_time

            # Store the raw response for debugging/review
            batch_key = f"Frames_{self.frame_count - len(frame_batch)*self.sample_rate + self.sample_rate}_to_{self.frame_count}"
            self.response_history[batch_key] = f"Inference Time: {inference_time:.2f}s\nRaw Response:\n{response_text}"

            self.logger.info(f"Batch inference time: {inference_time:.2f}s")
            # self.logger.debug(f"Raw response for batch: {response_text}") # DEBUG

            # Parse the single JSON response for the batch
            self._parse_json_response(response_text)

            self.logger.info(f"Batch Analysis - Classification: {self.last_prediction}, "
                             f"Confidence: {self.last_confidence}, Details: {self.last_details}")

            # Log alerts for concerning events based on batch analysis
            if self.last_prediction not in ["NORMAL", "INITIALIZING", "SYSTEM ERROR"]:
                log_level = logging.WARNING if self.last_confidence == "Low" else logging.ERROR
                self.logger.log(
                    log_level,
                    f"ALERT (Batch Analysis) - {self.last_prediction} detected with {self.last_confidence} confidence: {self.last_details}. Action: {self.last_action}"
                )

        except Exception as e:
            self.logger.exception(f"Error processing frame batch: {str(e)}")
            self.last_prediction = "SYSTEM ERROR"
            self.last_confidence = "High"
            self.last_details = f"Failed during batch processing: {str(e)}"
            self.last_action = "Investigate system logs."


    def _simulate_batch_ai_response(self, frame_batch):
        """Simulate AI response for a batch when Ollama isn't available"""
        # Very basic simulation: Look for significant change between first and last frame
        if len(frame_batch) < 2:
             return '{"primary_classification": "NORMAL", "confidence_level": "Low", "detailed_description": "Sequence too short for meaningful analysis.", "recommended_action": "Provide longer sequence."}'

        first_frame_gray = cv2.cvtColor(frame_batch[0], cv2.COLOR_BGR2GRAY)
        last_frame_gray = cv2.cvtColor(frame_batch[-1], cv2.COLOR_BGR2GRAY)
        first_frame_gray = cv2.GaussianBlur(first_frame_gray, (21, 21), 0)
        last_frame_gray = cv2.GaussianBlur(last_frame_gray, (21, 21), 0)

        frame_diff = cv2.absdiff(first_frame_gray, last_frame_gray)
        _, thresholded = cv2.threshold(frame_diff, 30, 255, cv2.THRESH_BINARY)
        changed_pixels = np.count_nonzero(thresholded)
        total_pixels = thresholded.shape[0] * thresholded.shape[1]
        change_percent = (changed_pixels / max(total_pixels, 1)) * 100

        self.logger.info(f"Simulation: Change between first and last frame: {change_percent:.2f}%")

        if change_percent > 15:
            desc = f"Significant change ({change_percent:.1f}%) detected across the frame sequence. Potential event occurred."
            rec_action = "Review sequence carefully."
            cls = "SUSPICIOUS" # Default suspicious for significant change
            conf = "Medium"
            # Add random chance for other events for variety in simulation
            rand_val = np.random.rand()
            if rand_val > 0.8: cls="VIOLENCE"; conf="Low"; desc += " Possible conflict detected."
            elif rand_val > 0.6: cls="ACCIDENT"; conf="Low"; desc += " Motion consistent with potential accident."

            return f'{{"primary_classification": "{cls}", "confidence_level": "{conf}", "detailed_description": "{desc}", "recommended_action": "{rec_action}"}}'
        elif change_percent > 5:
             return '{"primary_classification": "NORMAL", "confidence_level": "Medium", "detailed_description": "Moderate change detected, likely normal activity.", "recommended_action": "Continue monitoring."}'
        else:
             return '{"primary_classification": "NORMAL", "confidence_level": "High", "detailed_description": "Minimal change detected across the sequence, appears stable.", "recommended_action": "No action required."}'


    def _parse_json_response(self, response_text):
        """Parse the JSON response from the LLM."""
        try:
            # Attempt to clean potential markdown/formatting issues
            if response_text.strip().startswith("```json"):
                response_text = response_text.strip()[7:]
            if response_text.strip().endswith("```"):
                response_text = response_text.strip()[:-3]

            response_data = json.loads(response_text)

            # Validate using Pydantic model if structure is known
            # parsed = VideoMonitoringResponse(**response_data)
            # self.last_prediction = parsed.primary_classification
            # self.last_confidence = parsed.confidence_level
            # self.last_details = parsed.detailed_description
            # self.last_action = parsed.recommended_action

            # --- OR --- Manual parsing if Pydantic model isn't used/doesn't match
            self.last_prediction = response_data.get("primary_classification", "SYSTEM ERROR").upper()
            self.last_confidence = response_data.get("confidence_level", "Low")
            self.last_details = response_data.get("detailed_description", "No description provided.")
            self.last_action = response_data.get("recommended_action", "Review manually.")
            # --- End Manual Parsing ---

            self.logger.info(f"Parsed response - Classification: {self.last_prediction}, Confidence: {self.last_confidence}")

        except json.JSONDecodeError as e:
            self.logger.error(f"Failed to decode JSON response: {e}")
            self.logger.error(f"Problematic response text: {response_text}")
            self.last_prediction = "SYSTEM ERROR"
            self.last_confidence = "Low"
            self.last_details = f"Failed to parse JSON response. Raw: {response_text[:200]}..."
            self.last_action = "Check model output format."
        except Exception as e: # Catch other potential errors like pydantic validation
            self.logger.error(f"Error parsing response content: {str(e)}")
            self.last_prediction = "SYSTEM ERROR"
            self.last_confidence = "Low"
            self.last_details = f"Error interpreting response fields. Raw: {response_text[:200]}..."
            self.last_action = "Check model response structure."

    def _add_reading_info_overlay(self, frame, frame_num):
        """Add minimal info during the frame reading phase."""
        height, width = frame.shape[:2]
        font = cv2.FONT_HERSHEY_SIMPLEX
        text_color = (255, 255, 255)
        bg_color = (0, 0, 0)
        cv2.putText(frame, f"Reading Frame: {frame_num}",
                    (10, height - 10), font, 0.6, text_color, 1, cv2.LINE_AA)

    def _add_final_info_overlay(self, frame):
        """Add the final batch analysis information overlay to a frame."""
        height, width = frame.shape[:2]
        overlay = frame.copy()
        cv2.rectangle(overlay, (0, height - 120), (width, height), (0, 0, 0), -1)
        cv2.addWeighted(overlay, 0.7, frame, 0.3, 0, frame)

        font = cv2.FONT_HERSHEY_SIMPLEX
        text_color = (255, 255, 255)

        status_color = {
            "NORMAL": (0, 255, 0), "SUSPICIOUS": (0, 255, 255), "ACCIDENT": (0, 165, 255),
            "VIOLENCE": (0, 0, 255), "FIRE": (0, 0, 200), "THEFT": (255, 0, 255),
            "DISTURBANCE": (255, 255, 0), "MEDICAL": (255, 128, 0), "FIGHT": (0, 0, 255), # Fight same as violence
            "SYSTEM ERROR": (128, 128, 128), "OTHER": (200, 200, 200), "INITIALIZING": (100, 100, 100)
        }.get(self.last_prediction, (255, 255, 255))

        # Display batch analysis result
        cv2.putText(frame, f"Analysis Result (Batch)",
                    (10, height - 90), font, 0.6, text_color, 1, cv2.LINE_AA)
        cv2.putText(frame, f"Status: {self.last_prediction}",
                    (10, height - 60), font, 0.7, status_color, 2, cv2.LINE_AA)
        cv2.putText(frame, f"Confidence: {self.last_confidence}",
                    (width - 200, height - 60), font, 0.7, text_color, 1, cv2.LINE_AA) # Position confidence right

        # Display details (wrap text if too long)
        details_text = f"Details: {self.last_details}"
        max_len = width // 8 # Estimate characters per line
        if len(details_text) > max_len:
             # Simple wrap: find last space before max_len
             wrap_index = details_text.rfind(' ', 0, max_len)
             if wrap_index == -1: wrap_index = max_len # Force break if no space
             line1 = details_text[:wrap_index]
             line2 = "  " + details_text[wrap_index:].strip() # Indent second line
             if len(line2) > max_len: line2 = line2[:max_len-3] + "..." # Ellipsis if still too long
             cv2.putText(frame, line1, (10, height - 35), font, 0.5, text_color, 1, cv2.LINE_AA)
             cv2.putText(frame, line2, (10, height - 15), font, 0.5, text_color, 1, cv2.LINE_AA)
        else:
             cv2.putText(frame, details_text, (10, height - 30), font, 0.6, text_color, 1, cv2.LINE_AA)


    def _apply_final_overlay_to_output(self):
        """Applies the single final analysis overlay to all stored output frames."""
        if not self.output_frames:
            return
        self.logger.info("Applying final analysis overlay to output frames...")
        for frame in self.output_frames:
            self._add_final_info_overlay(frame)


    def _display_results(self):
        """Display video processing results (animation) in Colab."""
        if not self.output_frames:
            self.logger.warning("No output frames available to display.")
            return

        from IPython.display import HTML
        import matplotlib.pyplot as plt
        from matplotlib import animation
        import base64 # Already imported, but good practice

        self.logger.info(f"Creating results visualization with {len(self.output_frames)} frames...")

        # Create an animation
        if len(self.output_frames) > 1:
            self.logger.info("Creating animation of processed frames...")
            try:
                rgb_frames = [cv2.cvtColor(frame, cv2.COLOR_BGR2RGB) for frame in self.output_frames]
                fig = plt.figure(figsize=(10, (10 / rgb_frames[0].shape[1]) * rgb_frames[0].shape[0] )) # Adjust size based on aspect ratio
                plt.axis('off')
                im = plt.imshow(rgb_frames[0])

                def update(i):
                    im.set_array(rgb_frames[i])
                    return [im]

                interval_ms = max(200, 1000 / self.fps * self.sample_rate) # Rough estimate based on sample rate
                anim = animation.FuncAnimation(fig, update, frames=len(rgb_frames), interval=interval_ms, blit=True)
                plt.close(fig) # Prevent static plot from showing

                # Display the animation
                display(HTML(anim.to_jshtml()))
                self.logger.info("Animation displayed successfully.")
            except Exception as e:
                self.logger.error(f"Error creating animation: {str(e)}")
                # Fallback: Display first and last frames
                if len(self.output_frames) >= 2:
                    plt.figure(figsize=(15, 6))
                    plt.subplot(1, 2, 1)
                    plt.imshow(cv2.cvtColor(self.output_frames[0], cv2.COLOR_BGR2RGB))
                    plt.title("First processed frame")
                    plt.axis('off')
                    plt.subplot(1, 2, 2)
                    plt.imshow(cv2.cvtColor(self.output_frames[-1], cv2.COLOR_BGR2RGB))
                    plt.title("Last processed frame (with final analysis)")
                    plt.axis('off')
                    plt.tight_layout()
                    plt.show()
        elif len(self.output_frames) == 1:
             # Display single frame
             self.logger.info("Displaying the single processed frame.")
             plt.figure(figsize=(10, 6))
             plt.imshow(cv2.cvtColor(self.output_frames[0], cv2.COLOR_BGR2RGB))
             plt.title("Processed Frame (with final analysis)")
             plt.axis('off')
             plt.show()

        # Log final summary
        results_summary = {
            "total_frames_in_video": self.total_frames_in_video,
            "frames_sampled_for_analysis": len(self.frame_buffer), # How many actually sent
            "output_frames_generated": len(self.output_frames), # Frames for display
            "final_prediction": self.last_prediction,
            "final_confidence": self.last_confidence,
            "final_details": self.last_details,
            "final_action": self.last_action,
            "processing_time_seconds": self.last_process_time,
        }
        self.logger.info(f"Processing summary: {results_summary}")
        # Optionally display summary in Colab output
        from pprint import pprint
        print("\n--- Processing Summary ---")
        pprint(results_summary)
        print("\n--- Response History (Raw) ---")
        pprint(self.response_history)


# Example usage in Colab for the Batch Inferencer
def main_colab_batch():
    """Main function for Google Colab using Batch Inferencer"""
    print("Batch Video Inferencer for Google Colab")
    print("======================================")
    print("This version reads the whole video, samples frames,")
    print("and sends them in ONE batch to the model for analysis.")
    print("Processing happens AFTER the video is read.")

    os.makedirs("logs", exist_ok=True)
    # Frames saving is less relevant here as analysis is per-batch
    # os.makedirs("frames", exist_ok=True)

    # --- Configuration ---
    # MODEL_NAME = "llava:13b" # Or "llava:7b", "bakllava", etc. Check OllamaHub
    MODEL_NAME = "gemma3:12B" # Using 7b as it might be faster/less resource intensive
    SAMPLE_RATE = 5 # Sample 1 frame every 10 frames of the original video
    # FRAMES_PER_BATCH = 10 # This isn't used anymore as we send all sampled frames
    # --- End Configuration ---


    # Ask user for video source
    import ipywidgets as widgets
    from IPython.display import display as ipy_display # Avoid conflict with display function

    source_options = widgets.RadioButtons(
        options=['Upload a video file'], # Simplified for batch processing focus
        description='Video source:',
        disabled=False
    )
    ipy_display(source_options)

    # Create inferencer
    inferencer = ColabBatchVideoInferencer(
        model_name=MODEL_NAME,
        sample_rate=SAMPLE_RATE,
        # frames_per_batch is handled internally now (sends all sampled)
        output_dir="logs"
    )

    # Button to trigger upload and processing
    process_button = widgets.Button(
        description='Upload & Process Video',
        disabled=False,
        button_style='success',
        tooltip='Click to upload a video and start batch processing',
        icon='upload'
    )

    output_area = widgets.Output() # To capture prints and displays below the button

    def on_button_clicked(b):
        with output_area:
            clear_output() # Clear previous runs
            print("Button clicked. Initiating upload...")
            b.disabled = True
            b.description = 'Uploading...'
            uploaded = inferencer.upload_video()

            if uploaded:
                print(f"Video '{inferencer.video_source}' uploaded.")
                b.description = 'Processing...'
                print("Starting video analysis (reading frames first)...")
                try:
                    # Start processing - it will read, then analyze, then display
                    inferencer.start()
                    print("Analysis complete. Results should be displayed above.")
                except Exception as e:
                    print(f"\n--- ERROR during processing ---")
                    inferencer.logger.exception("An error occurred during the main processing flow.")
                    print(f"Error: {str(e)}")
                    print("Check the 'logs' directory for detailed logs.")
                finally:
                    b.description = 'Upload & Process Again'
                    b.disabled = False # Re-enable button
            else:
                print("Video upload failed or was cancelled.")
                b.description = 'Upload & Process Video'
                b.disabled = False # Re-enable button

    process_button.on_click(on_button_clicked)
    ipy_display(process_button, output_area)


# --- IMPORTANT ---
# To run this in Colab:
# 1. Make sure you have ollama installed *if* you are running Colab connected
#    to a local runtime where Ollama is running. Otherwise, it will use simulation.
#    (pip install ollama)
# 2. Run the cell containing the class definition.
# 3. Run the cell containing main_colab_batch().
# 4. Click the "Upload & Process Video" button that appears.
# 5. Upload your (preferably short) video file.
# 6. Wait for processing to complete. Results (animation, summary) will appear.

````
